**Purpose:**  
This notebook demonstrates how prompt design can guide a language model to act as a basic career advisor.  
The approach relies on structured role prompts, instructions, and example interactions, without using external datasets or APIs.


```python
# Role prompt
role_prompt = """
You are a student-friendly career guide.
Your responsibility is to give clear, simple, and supportive suggestions.
Avoid making up facts or giving unrealistic advice.
"""

# Instruction prompt
instruction_prompt = """
When responding:
- Suggest skill paths the student can follow
- Recommend beginner-friendly tools or topics
- Provide resume or portfolio guidance when needed
- Keep responses short and structured
"""

# Few-shot examples
few_shot_examples = """
User: I am in first year. How do I start learning technical skills?
AI:
- Begin with Python basics
- Practice small coding exercises daily
- Explore fields like cloud, data, AI, or cyber
- Build tiny projects to gain confidence

User: I want to prepare my resume. What should I add?
AI:
- Keep it simple and one page long
- Add current education and relevant skills
- Include 1–2 small academic or personal projects
"""
```

```python

def mock_llm(user_text):
    t = user_text.lower()

    # AI / ML path
    if any(word in t for word in ["ai", "ml", "aiml", "machine learning"]):
        return (
            "AI/ML Path (Medium Level):\n"
            "- Learn Python fundamentals and basic statistics\n"
            "- Understand data preprocessing and feature engineering\n"
            "- Practice model-building using Scikit-learn\n"
            "- Explore ML topics like classification, clustering, regression\n"
            "- Build small ML projects (spam classifier, house price prediction)"
        )

    # Data Analytics path
    if any(word in t for word in ["data", "analyst", "analytics"]):
        return (
            'Data Analytics Roadmap:\n'
            '- Learn Excel and SQL for data handling\n'
            '- Explore Python libraries like Pandas and Matplotlib\n'
            '- Practice creating dashboards using Power BI or Tableau\n'
            '- Work on small datasets (sales, weather, stock trends)\n'
            '- Present dashboards with clear insights'
        )

    # Cloud path
    if any(word in t for word in ["cloud", "aws", "azure", "gcp"]):
        return (
            "Cloud Computing Path:\n"
            "- Learn basics of servers, storage, virtualization\n"
            "- Understand how AWS EC2, S3, IAM work\n"
            "- Learn Docker basics (containers)\n"
            "- Deploy a simple website on cloud (free-tier)"
        )

    # Cybersecurity path
    if any(word in t for word in ["cyber", "security", "hacking"]):
        return (
            "Cybersecurity Starting Guide:\n"
            "- Learn networking fundamentals (IP, ports, protocols)\n"
            "- Understand Linux basics and command line usage\n"
            "- Explore beginner labs on TryHackMe\n"
            "- Learn the OWASP Top 10 vulnerabilities"
        )

    # Software Development path
    if any(word in t for word in ["developer", "software", "coding", "programming"]):
        return (
            "Software Development Track:\n"
            "- Choose one language: Python, Java, or C++\n"
            "- Practice Data Structures & Algorithms\n"
            "- Learn Git and GitHub for version control\n"
            "- Build small applications (calculator, login system, APIs)"
        )

    # Resume
    if any(word in t for word in ["resume", "cv"]):
        return (
            "Resume Tips:\n"
            "- Keep it one page and clean\n"
            "- Add technical skills clearly\n"
            "- Write 2–3 academic or personal projects\n"
            "- Highlight courses or certifications you completed"
        )

    # CSE
    if "cse" in t:
        return (
            "CSE Career Suggestions:\n"
            "- Strengthen programming + DSA\n"
            "- Explore domains like AI/ML, Cloud, Cybersecurity, Web Dev\n"
            "- Build GitHub projects\n"
            "- Learn SQL + one frontend or backend technology"
        )

    # ECE
    if "ece" in t:
        return (
            "ECE Career Options:\n"
            "- Software: Learn Python/Java + DSA\n"
            "- Electronics: Embedded systems, IoT, MATLAB basics\n"
            "- Consider hybrid roles (IoT, automation engineering)\n"
            "- Build small practical projects"
        )

    # Default fallback
    return (
        "General Career Advice:\n"
        "- Decide a domain you're comfortable with\n"
        "- Learn fundamentals consistently\n"
        "- Build small but real projects\n"
        "- Focus on improving your resume and portfolio"
    )

```

```python
def career_advisor(user_question):
    # Builds the complete structured prompt (required for assignment)
    complete_prompt = (
        role_prompt + "\n"
        + instruction_prompt + "\n"
        + few_shot_examples + "\n"
        + "User: " + user_question + "\nAI:"
    )

    # Response is generated by analyzing only the user question
    return mock_llm(user_question)
```

```python
user_query = input("Ask your career question: ")
print("\nAI Response:\n")
print(career_advisor(user_query))
```

```python
queries = [
    "I am a CSE AIML student. What should I learn?",
    "Is data analytics a good career?",
    "How do I begin with cloud computing?",
    "I want to improve my resume.",
    "Is cybersecurity a good option?",
    "I am from ECE. What path can I follow?"
]

for q in queries:
    print("USER:", q)
    print("AI:", career_advisor(q))
    print("-" * 50)
```

## Before vs After Prompt update

### Before update
Initially, the responses generated by the model were generic and repetitive.  
Even when different questions were asked, the output looked similar and lacked technical depth.  
The early prompt had the following limitations:

- The model did not clearly understand its role as a career advisor.
- The guidance was short, vague, and not tailored to the student’s background.
- Technical suggestions were either missing or too basic.
- No separation between domains (AI, Data, Cloud, Cybersecurity, etc.)
- Responses sometimes drifted away from the actual question.

Overall, the model behaved like a general assistant instead of a structured career guide.

### After update
After improving the structure of the prompt using **role definition**, **clear instructions**, and **well-designed examples**, the model's behavior improved significantly.  
The refined prompt now:

- Produces focused, domain-specific guidance
- Gives different answers for different queries
- Uses technical terminology clearly
- Provides practical steps such as tools, skills, and beginner projects
- Avoids unnecessary or unrelated information
- Follows a consistent structure that is easy to read

With these refinements, the model is now able to act as a reliable and supportive career advisor for technical students, providing advice that is both actionable and relevant.
